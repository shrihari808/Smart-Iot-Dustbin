```python
import cv2
import torch
from ultralytics import YOLO

# Load the YOLO model
model = YOLO(r"C:\Users\kshri\Desktop\codestuff\Tarp project\OUTPUT\runs\classify\train\weights\best.pt")

# Open the webcam
cap = cv2.VideoCapture(0)

# Set the desired resolution (e.g., 1280x720)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

# Capture a single frame
ret, frame = cap.read()

# Release the webcam
cap.release()

if ret:
    # Run the model on the captured frame
    results = model.predict(source=frame)

    # Create a window and bring it to the front
    window_name = "YOLO Predictions"
    cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
    cv2.setWindowProperty(window_name, cv2.WND_PROP_TOPMOST, 1)

    # Display the result frame with bounding boxes (if any)
    cv2.imshow(window_name, frame)

    # Wait indefinitely for the Escape key (ASCII 27) to close the window
    while True:
        key = cv2.waitKey(1)
        if key == 27:  # 27 is the Escape key
            break

    # Clean up
    cv2.destroyAllWindows()
else:
    print("Failed to capture image")
```

SERVO MOTOR

```python
import cv2
import serial
import time
import torch
from ultralytics import YOLO

model = YOLO(r"C:\Users\kshri\Desktop\codestuff\Tarp project\OUTPUT\runs\classify\train\weights\best.pt")

# Setup serial communication with Arduino
arduino = serial.Serial('COM3', 9600)  # Replace 'COM4' with your Arduino's COM port
time.sleep(3)  # Wait for the connection to be established

# Open the webcam
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

# Capture a single frame
ret, frame = cap.read()
cap.release()

if ret:
    results = model.predict(source=frame, show=False)

    if results is not None:
        detected = False
        for result in results:
            if result.boxes is not None:
                for detection in result.boxes.data.tolist():
                    class_id = int(detection[5])
                    confidence = detection[4]
                    
                    # Update: Check if item is recyclable with confidence > 0.7
                    if class_id == 0 and confidence > 0.7:
                        detected = True
                        break

        # Output based on detection and confidence threshold
        output = 1 if detected else 0
        print(f"Detected: {detected}, Sending: {output}")

        # Send output to Arduino
        arduino.write(str(output).encode())

        # Debug: Wait for Arduino response
        while arduino.in_waiting == 0:
            pass
        response = arduino.readline().decode('utf-8').strip()
        print(f"Arduino Response: {response}")

        # Display result frame
        cv2.imshow("YOLO Predictions", frame)
        cv2.waitKey(0)
        arduino.close()
        cv2.destroyAllWindows()
    else:
        print("No results returned from the model.")
else:
    print("Failed to capture image")
```
